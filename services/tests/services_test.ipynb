### recs_service

```python
# recommendations_online

recommendations_url = 'http://127.0.0.1:8000'

headers = {'Content-type': 'application/json', 'Accept': 'text/plain'}
params = {'user_id': 1337055}

resp = requests.get(f'{recommendations_url}/recommendations_online', headers=headers, params=params)

if resp.status_code == 200:
    recs = resp.json()
else:
    recs = []
    print(f'status code: {resp.status_code}')
    
print(recs) 
```

### events_service

```python
events_store_url = 'http://127.0.0.1:8020'

headers = {'Content-type': 'application/json', 'Accept': 'text/plain'}
params = {'user_id': 1337055, 'k': 100}

resp = requests.get(f'{events_store_url}/events', headers=headers, params=params)
if resp.status_code == 200:
    result = resp.json()
else:
    result = None
    print(f'status code: {resp.status_code}')
    
result
```
